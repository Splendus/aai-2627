## Lists

Lists are a very essential data type in Python. 

Normally and across programming languages we think of arrays as a *fixed size* sequence of values of *one type*. 

Python lists, on the other hand, can contain any type of variable and are dynamically extended. 

```python
some_list = ["a string", 1, 2]
print(some_list)
some_list.append(4)   # appends integer 4 in place
print(some_list)
```

Python does have an `array` module which enforces a type and is more lightweight in terms of memory, but it is very rarely used. Much more frequently used, and what we'll soon focus on, are `numpy` arrays. These come closer to the common interpretation of arrays having a fixed size and dtype.

Lists can also be iterated over in a straightforward manner:

```python
number_list = []  # creates an empty list and is equivalent to `number_list = list()`
number_list.append(1)  # add elements at the end of the list
number_list.append(2)
number_list.append(3)
print(number_list)
print("The first element of the number list is: ", number_list[0])  # accessing a single list element by its index
print("The third element of the number list is: ", number_list[2])

print("Looping over the list:")
for number in number_list:
    print(number)
```

Accessing an index that doesn't exist generates an exception (``IndexError``):

```python
number_list = [1, 2, 3]
print(number_list[9])
```

### Negative Indexing
With negative indexing it is possible to count the list elements starting with -1 from the list's tail.

![Negative Indexing](https://developers.google.com/edu/python/images/hello.png)

```python
colors = ['red', 'green', 'blue', 'yellow', 'white', 'black']
print('Last element:', colors[-1])
print('Penultimate element:', colors[-2])

print()
# Which is the same as subtracting the negative index from the lists total lenght
print('Last element:', colors[len(colors)-1])
print('Penultimate element:', colors[len(colors)-2])
```

### List Slicing
Slicing a list is used when you only want to retrieve a part (a *slice*) of the list. The basic slicing syntax is ``start:stop``, with ``start`` being inclusive, ``stop`` being exlusive.

```python
numbers = [10, 20, 30, 40, 50, 60, 70, 80, 90]

# take elements with index 2 to 6
print(numbers[2:7])
```

The full slicing syntax is ``start:stop:step``. ``step`` allows us to only take each nth element from the list:

```python
# take every 2nd element
numbers[2:7:2]
```

If ``start`` is the first element or ``stop`` is the last, you can drop the index:

```python
# take the first 5 elements
print(numbers[:5])

# take all elements beginning from index 5
print(numbers[5:])
```

Note that all the indexing methods showed above also work for other `Sequence` types like strings (accessing characters) and tuples.

### List Methods
Given the following list

```python
l = [1, 2, 3]
```

##### Adding a single element to the end of a list with `append()`

```python
# Append is an in-place operation. It modifies (appends) the existing list
l.append(4)
print(l)
```

##### Adding sequence of elements to the end of a list with `extend()`

```python
l.extend([4, 5, 6])
print(l)
```

Alternatively, we can also use the `+` operator to concatenate two lists.

```python
l = [1, 2, 3, 4, 5, 6] + [7, 8, 9]
print(l)
```

##### Inserting an element at a specific position in the list with `insert()`

```python
l.insert(1, 'x')
print(l)
```

##### Removing (and returning) an element at a specific position in the list with `pop()`

```python
# Pop(<index>) removes and returns the element at position <index> from the list
l.pop(1)
```

```python
l
```

##### Remove the *first* element that has a given *value* (as opposed to index) with `remove()`

```python
l.extend([2,2,2])
print(l)

l.remove(2)
print(l)
```

##### Count how often a certain elements appears in the list with `count()`

```python
print(l.count(2))
```

##### Sort the list with `sort()`; Default in ascending order:

```python
l.sort()
l
```

```python
l.sort(reverse=True)
print(l)
```

### Membership checking

`list` does not provide a method to check whether a certain values is in a list. However, we can use the `in` statement for membership checking.

```python
# Check if 3 is in the list
3 in l
```

```python
# Check if 99 is in the list
99 in l
```

Of course, we can also check whether an element is not in the list using `not in`.

```python
# Check if 77 is not in the list
77 not in l
```

**Note that checking whether an element exists in a list has an average time complexity of O(n)!**

<div class="alert alert-block alert-warning">
<b>Caution:</b> Note that we cannot test for sub-sequence membership in other list (sequence).
</div>


```python
# This does NOT work!
[1, 2] in [1, 2, 3]
```

### How can we check whether two lists are equal?

We sometimes want to know whether two given lists are equal. If the order of the elements should be taken into account such check can be easily performed using the `==` operator.

```python
l = [1, 2]
```

```python
# The following lists should be equal
l == [1, 2]
```

```python
# The following lists should not be equal
l == [1, 2, 3]
```

```python
# The following lists should also not be equal
l == [3, 1]
```

```python
# The following lists should not be equal as well
l == [1]
```

But what if the order should not play a role in our check? <br />
Typically, we see the following two solutions being used in practice.

##### Method 1: using sort

```python
# Sort both lists and check whether they are equal
l1 = [1, 2]
l2 = [2, 1]

l1.sort()  # the `.sort()` works in-place
l2.sort()

l1 == l2
```

##### Method 2: using sets


```python
# Convert both lists into sets (unordered!) first and then compare both sets
l1 = [1, 2]
l2 = [2, 1]

print(set(l1) == set(l2))
```

However, there is one important thing that you should be aware of when using the second method. 
Let's look at the following example:

```python
# Convert both lists into sets (unordered!) first and then compare both sets
l1 = [1, 2]
l2 = [2, 1, 2]

set(l1) == set(l2)
```

The sets of the lists are equal while the lists (even if ordered) are not. This happens because **sets do not contain duplicates**.

### List unpacking

We can unpack lists into individual variables.

```python
l = [1, 2, 3]

# Unpack the list into three variables
x, y, z = l

print(x)
print(y)
print(z)
```

You can also use the unpacking operator `*` for similar tasks:

```python
print(*l)
```

The unpacking operator only works in certain contexts, such as function definitions and calls.

```python
x, y, z = *l
```
