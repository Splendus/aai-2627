## Basic Syntax
Python syntax is clear and easy-to-read. However, it has a strict style using indentation instead of braces.  ``#`` is used to mark single-lined comments.

__Source:__ https://docs.python.org/3.13/reference/lexical_analysis.html#line-structure

```python
# no semicolon at the end of a statement
x = 1
# colon at the end of conditions and loops
if x == 1:
    # indentation to denote different levels of scope
    print('x is one.')  # this comment ends a line
```

```python
# You could, however, put a semicolon at the end of a statement
y = 2;
# That way, you could put multiple statements in one line
x = 1; y = 2; z = 3;
# but usually this is considered bad practice and discouraged
# In this case, it would be better to give each variable one line
x = 1
y = 2
z = 3
# or do
x, y, z = 1, 2, 3
```

## Line Joining
There are two ways of joining lines in Python: implicit and explicit.

### Explicit line joining
Two or more physical lines may be joined into logical lines using backslash characters `\`, as follows: when a physical line ends in a backslash that is not part of a string literal or comment, it is joined with the following forming a single logical line, deleting the backslash and the following end-of-line character. For example:

```python
if 1900 < 2010 < 2100 and \
   1 <= 12 <= 31:
    print("This is an explicit line join!")
```

### Implicit line joining

Expressions in parentheses, square brackets or curly braces can be split over more than one physical line without using backslashes. Using the previous example:

```python
if (
    1900 < 2010 < 2100
    and
   1 <= 12 <= 31
):
    print("This is an implicit line join!")
```

Or with square brackets

```python
month_names = ['Januari', 'Februari', 'Maart',      # These are the
               'April',   'Mei',      'Juni',       # Dutch names
               'Juli',    'Augustus', 'September',  # for the months
               'Oktober', 'November', 'December']   # of the year

print(month_names)
```

Implicitly continued lines can carry comments. The indentation of the continuation lines is not important. Blank continuation lines are allowed. There is no NEWLINE token between implicit continuation lines. Implicitly continued lines can also occur within triple-quoted strings `""" """` (see below); in that case they cannot carry comments.

## Variables and Types
Python is dynamically typed. In contrast to statically typed languages, variables (and their types) do not need to be declared before using them.

### Numbers
Python natively supports three types of numbers: integers, floating point numbers and complex numbers. Here, we focus on the first two.

#### Integers


```python
int_number = 7
print(int_number)
```

The default is integer

```python
print(type(-123))
print(type(123))
```

Python3 integer values do not have a max limit. We can handle values as large as the available memory allows.

Internally, Python 3 stores every integer as an arbitrary-precision number, which simply takes up more memory as it grows.

```python
print(10**1000)
```

#### Floats
Floats are identified by a period `.` in their definition

```python
float_number = 7.0
print(float_number)
print(type(float_number))
```

They can also be *typecasted* from an integer via `float()`

```python
float_number = float(7)
print(float_number)
print(type(float_number))
```

Floats are represented as 64-bit double-precision (type double in C) values. Floating point numbers can be in the range of $-1.7976931348623157 \times 10^{308}$ and $1.7976931348623157 \times 10^{308}$.

We can verify this using the `sys` module.

```python
import sys
print(sys.float_info.max)
```

If we further increase the number, it becomes infinity.

```python
print(sys.float_info.max * 1.1)
```

The float of smallest magnitude is $2.2250738585072014 \times 10^{-308}$

```python
print(sys.float_info.min)
```

You could potentially go below that (into the *subnormal* range) but this would sacrifice precision

```python
smallest_rep_num = sys.float_info.min * sys.float_info.epsilon
print(smallest_rep_num)
print("The next representable smaller number is already: ", smallest_rep_num / 2)
```

### Strings
Strings are python's data structure for text data. They are defined by using either a single quote or double quotes:

```python
print('hello')
print("hello")

assert 'hello' == "hello"  # assert will throw an error if the following statement is evaluated to False
```

The difference between the two is that using double quotes makes it easy to include apostrophes. (These would terminate the string if using single quotes.)

```python
apostrophe_sent = "Don't worry about apostrophes."
print(apostrophe_sent)
```

A backslash can be added at the end of a line to ignore the newline character:

```python
'This string will not include \
backslashes or newline characters.'
```

Or with three single (''') or double quotes (""") if we want to define a multi-line string without need for breaking lines via `\`. Usually, three quotes are used to document functions and classes.

```python
s1 = """Line 1
Line 2"""


print(s1)
```

Simple operators can be executed on numbers and strings:

```python
one = 1
two = 2
three = one + two
print(three)

hello = "hello"
world = "world"
helloworld = hello + " " + world
print(helloworld)
```

Python is *strongly typed*, which leads to the mixing of operators between numbers and strings to not be supported (``TypeError``):

```python
# This will throw a TypeError
one = 1
two = 2
hello = "hello"
print(one + two + hello)
```

Note that, unlike in Python2, in Python3 all strings are **unicode** strings by default. <br />
Hence, we can write text that contains characters such as "ü" or "ä" out-of-the-box.

```python
print('Hüllo')
print('à')
```

If we explicitly need a binary string for some task, we can explicitly enforce this format by adding a "b" upfront. In byte strings, every element is a single byte, i.e. a value between 0 and 255. However, only ASCII characters (values 0-127) can be written directly; bytes with a value of 128 or greater must be written as escape sequences, e.g. `b'\xff'` for the value 255.

```python
b_string = b'This is a binary string'
print(b_string)
```

```python
print(b'This is a binary string with ä')
```

### A note on typing
While python is indeed a dynamic language, it does support type *hints*. This is seldom used outside of function documentation but may serve a similar purpose as commenting your code. Types are hinted at by a colon `:` after the variable name.
> Note: Python's interpreter and also any variant of its JIT compilers will *ignore* your type hints. They can not be used to speed up code by skipping some type checks. This also means that you can use type hints in a confusing way. They are merely a gadget to make your code more verbal which might be good in some contexts but can also be detrimental.

```python
i: int = 1  # makes sense

some_string: str = [1, 2, 3]  # rather confusing, but neither a warning nor an error will be raised
```
