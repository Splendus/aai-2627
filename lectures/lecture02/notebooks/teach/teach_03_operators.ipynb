**QUESTION**

How to appropriately round numbers?

```python
print(0.5, round(0.5))
print(1.5, round(1.5))
print(2.5, round(2.5))
```

As can be seen, both 1.5 and 2.5 are rounded to 2. This happens because `round()` applies scientific rounding (*round half to even*): numbers that lie exactly halfway between two integers are rounded toward the **nearest even** number. All other numbers are rounded to the nearest integer as usual, e.g. `round(2.6)` gives `3`. Hence, it's different from what we call "Kaufmännisches Runden".

```python
# Example: "Kaufmännisches Runden"
# Note: this trick only works for positive numbers, e.g. int(-1.5+r_const) gives -1 instead of -2
r_const = 0.5
print(int(0.5+r_const))
print(int(1.5+r_const))
print(int(2.5+r_const))
print(int(0.4+r_const))
print(int(1.4+r_const))
print(int(2.4+r_const))
```

### Bitwise operators

Bitwise operators can be used to perform bitwise comparisons of two values. <br/>
The following bitwise operations are available: `&`, `|`, `~`, `^`, `>>`, `<<`

```python
# Left shift
# x = 0000 0001
x = 1
# Left shift by 1. Should become 0000 0010 ==> 2
print(x << 1)
# Left shift by 3. Should become 0000 1000 ==> 8
print(x << 3)
```

```python
# Bitwise NOT. ~ Operators. Inverts every bit
x = 0

print(~x)
```

**Why do we get -1 if we invert 0?** <br/>
Python assumes that the numbers are represented using two's complement. See: https://en.wikipedia.org/wiki/Two%27s_complement

Assuming that 0 is represented as an 8-bit number (0000 0000), we get 1111 1111 after inverting the bits.<br/>
In two's complement, the first (left) bit has the negative weight $-2^7$, while the remaining seven bits add up to $2^7-1$. Hence, we get: <br/>

$$-2^7 + (2^7-1) = -1$$

```python
# Values can be bitwise combined using the &, |, ~ and ^ (XOR) operator.
x = 5   # 0101
y = 12  # 1100
```

```python
# AND: Should yield 0100 ==> 4
print(x & y)
```

```python
# OR: Should yield 1101 ==> 13
print(x | y)
```

In practice, we sometimes find bitwise operators being used with booleans. 

```python
print(True & True)
print(True & False)
print(True | False)
```

This is possible because `bool` is a subtype of `int`: `True` is just the integer `1` and `False` the integer `0` (see below). For these two values, bitwise and logical operators give the same result. <br/>
Careful: this does not carry over to other numbers. Although every non-zero number counts as `True` in a condition, `bool(2 & 1)` is `False`, whereas `bool(2 and 1)` is `True`.

```python
# We can verify this ourselves by casting the values to int.
print(int(True))
print(int(False))
```

```python
print(bool(2))
print(bool(0))
```

**Why is it still better to use logical operators if we want to evaluate conditional expressions?**

Because logical operators are **short-circuit operators** which enable faster evaluation.

**Example:** <br/>
Let's assume that we have the following expression:

`False & <Expression>` 

There is no need to evaluate the entire expression once we have noticed that the first term (`False`) is already false. 

If we use bitwise operations, short-circuit evaluation is impossible. The entire expression needs to be evaluated.

### Identity Operators

Identity operators are used to compare the objects, *not if their values are equal*, but if they are actually the same object with the same memory location.

Operators: `is`, `is not`


```python
x = 3
y = 4
print(x is x)
print(x is y)
```

```python
# We can get the memory location of a variable using id()
print(id(x))
print(id(y))
```

Interestingly, the following code evaluates to `True`. Why?

```python
x = 3
y = 3
print(x is y)
```

```python
# Let's look at the memory location
print(id(x))
print(id(y))
```

Apparently, both 3s refer to the same object (recall that in Python everything is an object)! <br/>
In fact, x and y are names (a.k.a. references) that refer to the same object. This is an optimization of the standard Python interpreter (CPython): small integers from -5 to 256 are created only once and then reused. For larger numbers, e.g. `x = 1000` and `y = 1000` in two separate cells, `x is y` gives `False`. Hence, always compare values with `==`.

### Operator Precedence

Finally, let's take a brief look at the operator precedence, which defines the order in which expressions are evaluated.

See: https://www.w3schools.com/python/python_operators.asp

As can be seen, comparison operators (e.g. `>`, `<`, `==`) rank higher than the logical operators `not`, `and` and `or`. Hence, we can omit the brackets in expressions such as ...

```python
print((5 > 3) and (5 < 6))
print(5 > 3 and 5 < 6)
```

**Careful:** The bitwise operators `&`, `|` and `^` rank *higher* than comparisons. Hence, `x > 3 & x < 6` is evaluated as `x > (3 & x) < 6`. For `x = 7`, this gives `True`, although 7 is not smaller than 6. We will run into this again with pandas.
