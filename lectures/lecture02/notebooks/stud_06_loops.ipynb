## Loops
There are two types of loops in Python, ``for`` and ``while``. For loops work very intuitively and do not neccessarily require an incrementation of integers.

### The for Loop
<img src="https://swcarpentry.github.io/python-novice-inflammation/fig/loops_image.png" width="250">


```python
word = 'oxygen'
for char in word:
    print(char)
```

```python
primes = [2, 3, 5, 7]
for prime in primes:
    print(prime)
```

For loops can iterate over a sequence of numbers using the ``range`` function. The way ``range`` works is quite similar to list slicing.

Signature of the `range` function:
`range(<start>, <stop>, <step_size>)`
with `start` and `step_size` being optional. 

This means, if only one number is given to `range`, this number will function as the (exlusive) `end` argument.

```python
for x in range(5):
    print(x)
```

```python
for x in range(3, 6):
    print(x)
```

```python
for x in range(3, 8, 2):
    print(x)
```

```python
type(range(3, 8, 2))
```

For loops iterate over a given `Iterable`, that is, as the name implies, all types one could iterate over. 
Naturally, a `list` but other types can be iterated over as well.

> If certain objects are of a certain type, i.e. are an *instance* of a certain class can be checked with the very useful builtin function `isinstance(object, class)`

```python
print(isinstance(1, int))
print(isinstance(1.0, int))
```

### while Loops
While loops behave similarly as in other programming languages, repeating as long as a certain condition is met.

```python
count = 0
while count < 5:
    print(count)
    count += 1
```

### Break and Continue
`break` and `continue` statements are useful in contexts where termination of a loop is not known beforehand (or is determined lazily). In short, `break` will terminate the entire loop while `continue` will skip the current iteration.
