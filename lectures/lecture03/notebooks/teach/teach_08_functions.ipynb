## Functions

Functions in Python are defined using the keyword ``def`` followed by the function's name. By convention, function names are lower and snake case.

```python
def greeting_function():
    print("Greetings from this function.")
```

Naturally, Python functions can also take arguments and return values:

```python
def sum_two_numbers(a, b):
    print(f'The sum of {a} + {b} is {a + b}.')
    return a + b
```

Now, let's call the functions

```python
greeting_function()
result = sum_two_numbers(7, 3)
```

Note that all Python functions return some value. The return value is ``None`` if the return statement is omitted or the statement is just ``return``. It is also possible to return multiple values.

### Default Arguments
Python allows function arguments to have default values. If the function is called without the argument, the argument gets its default value. The default value is assigned by using the operator ``=``.

```python
def sum_two_numbers(a, b = 5):
    print(f'The sum of {a} + {b} is {a + b}.')
    return a + b


num_plus5 = sum_two_numbers(3)
print(num_plus5)
num_plus7 = sum_two_numbers(3, 7)
print(num_plus7)
```

When using more than one default argument, it adds to readability when including the argument name in the function call:

```python
def process_string(sentence, lowercase=False, split=False, strip=False):
    if lowercase:
        sentence = sentence.lower()
    if strip:
        sentence = sentence.strip()
    if split:
        sentence = sentence.split()
    return sentence


process_string('   How Python handles multiple default arguments:    ',
               lowercase=True,
               strip=True)
```

In function calls, *keyword* arguments (those parameters identified by their keyword) must follow *positional* arguments (those parameters identified by their position).

```python
process_string(lowercase=True,
               strip=True,
               '   How Python handles multiple default arguments:    ',)
```

Similarly for function definitions, parameters with default values have to follow those without one.

```python
def greet_person(greeting = "Hello", person):
    print(greeting, person)
```
