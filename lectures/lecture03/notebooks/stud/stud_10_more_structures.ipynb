## Tuples
Tuples are similar to lists but have an important distinction: They are *immutable*.

```python
tup = (1, 2, 3)

# Try to mutate tuple
tup[1] = 5
```

Like lists, they can also be indexed, sliced, nested, concatenated, etc.

```python
tup1 = (1, 2, 3)
tup2 = (4, 5)
tup_cat = tup1 + tup2
print(tup_cat)

tup_nest = (tup1, (tup1, tup2))
print(tup_nest)

tup_slc = tup_cat[:3]
print(tup_slc == tup1)  # values are the same
print(tup_slc is tup1)  # underlying objects are not
```

Although both lists and tuples can contain diverse data types, it is convention to have lists carry one type while tuples can carry more.

```python
a_tup = (1, 2, "John", 3.0)  # OK

a_lis = [1, 2, "John", 3.0]  # works but not encouraged
```

Tuples pop up in a lot of places in python, e.g. functions returning multiple values

```python
def a_func():
    return 1, 2, 3
return_vals = a_func()
print(type(return_vals))
```

Notoriously, tuples are indicated by a `,` and not necessarily by parentheses. That is, if you encounter random tuples in your code, it might be, you have a comma somewhere lying around where it should not be

```python
a_tup = (1, 2)
print(type(a_tup))

not_a_tup = (1)
print(type(not_a_tup))

surprisingly_a_tup = 1,
print(type(surprisingly_a_tup))
```

## Dictionaries
A dictionary is a data type that works with *keys* and *values*, it is in fact a *map* from *keys* to *values*. Each value stored in a dictionary can be accessed using its key instead of its index. A key can be any *hashable* (roughly: immutable) object, e.g. a string, a number or a tuple of those, but not a list. A dictionary is indicated by braces ``{}``.

Fill initially empty dictionary.

```python
mlb_team = {}  # same as mlb_team = dict()
mlb_team['Colorado'] = 'Rockies'


mlb_team['Boston'] = 'Red Sox'
mlb_team['Minnesota'] = 'Twins'
mlb_team['Milwaukee'] = 'Brewers'
mlb_team['Seattle'] = 'Mariners'
print(mlb_team)
```

One can also directly fill dictionary with items (key:value - pairs)

```python
mlb_team = {
    'Colorado': 'Rockies',
    'Boston': 'Red Sox',
    'Minnesota': 'Twins',
    'Milwaukee': 'Brewers',
    'Seattle': 'Mariners'
}
print(mlb_team)
```

Or with the `dict()` constructor and keyword arguments. The keys are then written without quotes, which only works for keys that are valid variable names (e.g. not `"New York"`).

```python
mlb_team = dict(
    Colorado = 'Rockies',
    Boston = 'Red Sox',
    Minnesota = 'Twins',
    Milwaukee = 'Brewers',
    Seattle = 'Mariners'
)
print(mlb_team)
```

### Accessing Dictionary Values
The dictionary entries are displayed in the order they were defined. But that is irrelevant when it comes to retrieving them. Dictionary elements are not accessed by numerical index:

```python
mlb_team[0]  # this throws an exception (unless there is an integer key 0)
```

The same error is raised when trying to access a nonexistent key:

```python
mlb_team['Pittsburgh']
```

Access a dictionary's values using keys. This way, you can also update a given value or delete it.

```python
print(mlb_team['Seattle'])
```

```python
mlb_team['Seattle'] = 'Pilots'
print(mlb_team['Seattle'])
```

```python
del mlb_team['Seattle']  # or: mlb_team.pop('Seattle')
print(mlb_team)
```

You can also iterate over all dictionary items:

```python
for city, team in mlb_team.items():
    print(f"{city}'s team is called {team}.")
```

Be aware of overriding keys:

```python
mlb_team["Chicago"] = "Cubs"
mlb_team["Chicago"] = "White Sox"

print(f"{'Cubs' in mlb_team.values()=}")
print(f"{'White Sox' in mlb_team.values()=}")
```

### Lists vs. Dictionaries
1. Both are mutable.
2. Both are dynamic. They can grow and shrink as needed.
3. Both can be nested: A list can contain another list. A dictionary can contain another dictionary. A dictionary can also contain a list, and vice versa.
4. List elements are accessed by their position via indexing. Dictionary elements are accessed via keys.

## Mutable vs Immutable Objects
Simply put, mutable objects can be changed after they are created, while immutable objects can't.

**Immutable objects:**

int, float, string, tuple, bytes

**Mutable objects:**

list, dict, set

We now create an object of type int. Note how ``x`` and ``y`` point to the same object. 


__CPython implementation detail:__ For CPython, `id(x)` is the memory address where `x` is stored.


```python
x = 10
y = x
print(f'{id(x)}, {id(y)}, {id(10)}')
```

compare the values

```python
print(x == y)
print(x == 10)
print(y == 10)
```

compare the objects (= object ids)

```python
print(x is y)
print(x is 10)
print(y is 10)
```

See what happens when we change the value of "x":

```python
x += 1
print(f'{id(x)}, {id(y)}, {id(10)}')
```

compare the values

```python
print(x == y)
print(x == 10)
print(y == 10)
```

compare the objects (= object ids)

```python
print(x is y)
print(x is 10)
print(y is 10)
```

"Changing the value" of an immutable object in fact **creates a new object**! Object 10 was never changed.

Now compare this to the behaviour of a mutable object, e.g. a list:

```python
first_list = [1, 2, 3]
also_first_list = first_list
print(f'{id(first_list)}, {id(also_first_list)}')
```

compare the values

```python
print(first_list == also_first_list)
```

compare the objects (= object ids)

```python
print(first_list is also_first_list)
```

And see what happens if we change the list (e.g. removing an element):

```python
first_list.pop()
print(first_list)
print(f'{id(first_list)}, {id(also_first_list)}')
```

compare the values

```python
print(first_list == also_first_list)
```

compare the objects (= object ids)

```python
print(first_list is also_first_list)
```

Both names still point to the same object (same object id). There is only one list, so the modification is visible through **both** names!

*from: [Blog post](https://medium.com/@meghamohan/mutable-and-immutable-side-of-python-c2145cf72747)*

```python
print(first_list)
print(also_first_list)
```

In other words, an object that allows you to **change its values without changing its identity** is a **mutable** object.

## Sets
A set has the following characteristics:
- Sets are unordered.

- Set elements are unique. Duplicate elements are not allowed.

- A set itself may be modified, hence is mutable, but the elements contained in the set must be of an immutable (technically *hashable*) type. (A set *cannot* contain sets but it *can* contain its immutable cousin `frozenset`)

So sets are neither *Sequences* (like lists, tuples, strings) nor *Mappings* (like dicts).

Let's say you want to collect a list of unique words used in a sentence. This is where a set comes in handy:

```python
print(set("my name is Jonathan and Jonathan is my name".split()))
```

## List Comprehension
List comprehensions are an advanced Python topic that are really useful to shorten code (whilst usually remaining readable).

Take a look at the code below and try to explain in words, what it does.

```python
sentence = "the quick brown fox jumps over the lazy dog"
words = sentence.split()
word_lengths = []
for word in words:
    if word != "the":
        word_lengths.append(len(word))
print(word_lengths)
```

Next, look at the code below and see how elegantly this can be achieved in a single line (instead of five):

```python
word_lengths = [len(word) for word in sentence.split() if word != "the"]
print(word_lengths)
```

### Exercise: Mutable default arguments

One proverb among python developers is that *mutable default arguments are the root of all evil*.
In this exercise we will consider this a bit further.
Consider the following function, that appends its first argument, a value, to its second argument, a list. By default, the second argument will be an empty list

```python
def append_to(value, value_list = []):
    value_list.append(value)
    return value_list
```

As expected 

```python
first_list = append_to(1)
print(f"{first_list=}") # expected [1]
```

But then on successive calls:

```python
second_list = append_to(2)
print(f"{second_list=}")  # expected [2]
```

That is, the function in some sense *tracks state*, i.e. the value of its default argument `value_list`.
The `value_list`'s state is not simply wiped by supplying a value for it as

```python
list_with_specified_value_list = append_to(13, value_list=[11, 12])
print(f"{list_with_specified_value_list=}")  # expected [11, 12, 13]
```

```python
third_list = append_to(3)
print(f"{third_list=}")  # expected [3]
```
